```
from datetime import date, timedelta

import requests
```

Choose a monitoring station. This one is Day Brook in Nottingham. Let's get the total rainfall for yesterday.

Note that the real-time API only keeps around four weeks of readings, so older dates return no data.

```
# stations are identified by a unique numeric identifier
station = 4163
# we need to provide the date in iso format
day = date.today() - timedelta(days=1)
```

The URL is created based on following the structure given in the official [documentation](https://environment.data.gov.uk/flood-monitoring/doc/reference#stations).

```
URL = (
    "https://environment.data.gov.uk/flood-monitoring/id/measures/"
    f"{station}-rainfall-tipping_bucket_raingauge-t-15_min-mm/readings"
)
```

The data is returned in JSON format. We can use the `requests` library to get the data and parse it into standard Python data structures. The built-in urllib library can also be used to get the data, but it is more verbose and less convenient. It's common to use the `requests` library for this purpose.

```
# note that we can use the params argument to pass query parameters
response = requests.get(URL, params={"date": day.isoformat()})
response.raise_for_status()  # check for issues
rainfall = response.json()  # get the rainfall as json
```

Finally, we can process the data to get the total rainfall for the day.

```
readings = rainfall["items"]
if not readings:
    raise ValueError(f"No readings for {station} on {day.isoformat()}.")

# the daily total is the sum of all readings
day_rain = sum(item["value"] for item in readings)

print(f"{day.isoformat()}: {day_rain:.1f} mm from {len(readings)} readings")
```
